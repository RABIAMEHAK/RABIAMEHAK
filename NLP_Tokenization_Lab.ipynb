
#  Tokenization Techniques
**Objective:** Explore and implement different tokenization techniques using Python, handle special cases, and analyze tokenization performance.

**Required Libraries:** `nltk`, `spacy`, `re`, `string`


##  Task 1: Display and Analyze the Text

```python

import nltk, re, string
from collections import Counter

nltk.download('punkt')

text = """Dr. Smith graduated from Harvard University in 2009. 
He is currently working at OpenAI developing AI-powered solutions. 
Email him at dr.smith@example.com or visit https://openai.com for details!"""

print("=== Original Text ===")
print(text)
```

##  Task 2: Word Tokenization using NLTK

```python

from nltk.tokenize import word_tokenize

tokens_nltk = word_tokenize(text)
print(tokens_nltk)

print("\nTotal Tokens:", len(tokens_nltk))
print("Unique Tokens:", len(set(tokens_nltk)))

freq = Counter(tokens_nltk)
print("\n=== Token Frequency ===")
for token, count in freq.items():
    print(f"{token}: {count}")
```

##  Task 3: Sentence Tokenization using NLTK

```python

from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(text)
print(sentences)
print("\nTotal Sentences:", len(sentences))

sent_lengths = [len(s.split()) for s in sentences]
longest = sentences[sent_lengths.index(max(sent_lengths))]
shortest = sentences[sent_lengths.index(min(sent_lengths))]

print("\nLongest Sentence:", longest)
print("Shortest Sentence:", shortest)
```

##  Task 4: Tokenization using Regular Expressions

```python

text_clean = re.sub(r'[^\w\s@.:/-]', '', text)
tokens_regex = re.findall(r'\b\w[\w.-]*\b', text_clean)

print(tokens_regex)
print("\nNLTK Token Count:", len(tokens_nltk))
print("Regex Token Count:", len(tokens_regex))
print("\nDiscussion:")
print(" Regex-based tokenization removed all punctuation marks completely.")
print(" However, it also removed special structures like emails and URLs,")
print("   which are better preserved by NLTK’s tokenizer.")
print(" NLTK handled punctuation and special tokens (like dr.smith@example.com, https://openai.com) more accurately.")
```

##  Task 5: Tokenization using SpaCy

```python
import spacy
from nltk.tokenize import word_tokenize
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
tokens_spacy = [token.text for token in doc]
print("SpaCy Tokens:")
print(tokens_spacy)
tokens_nltk = word_tokenize(text)
print("\nNLTK Tokens:")
print(tokens_nltk)
spacy_specials = [t.text for t in doc if "@" in t.text or "http" in t.text]
nltk_specials = [t for t in tokens_nltk if "@" in t or "http" in t]

print("\nSpaCy detected email/URL tokens as:")
print(spacy_specials)

print("\nNLTK detected email/URL tokens as:")
print(nltk_specials)

print("\nConclusion:")
print("SpaCy handled email and URL tokens more accurately than NLTK.")
print("SpaCy keeps them as single, meaningful tokens, while NLTK may split them into smaller parts.")
```

##  Task 6: Custom Tokenizer Function

```python
def custom_tokenizer(text):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    return text.split()

custom_tokens = custom_tokenizer(text)
nltk_tokens = word_tokenize(text)

print("Custom Tokens:", custom_tokens)
print("\nNLTK Tokens:", nltk_tokens)
print(f"\nCustom Tokenizer Tokens: {len(custom_tokens)}")
print(f"NLTK Tokens: {len(nltk_tokens)}")
print("\nCustom tokenizer is simple but removes punctuation and capitalization.")
```

##  Task 7: Subword Tokenization (Advanced)

```python

from nltk.tokenize import SyllableTokenizer

syl_tok = SyllableTokenizer()
words = ["processing", "tokenization", "university"]

for w in words:
    print(w, "->", syl_tok.tokenize(w))

print("\nSubword tokenization helps handle rare/unknown words and reduce vocabulary size.")
```
