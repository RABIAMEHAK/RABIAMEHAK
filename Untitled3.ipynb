```python
import nltk
from nltk.stem import WordNetLemmatizer, PorterStemmer
from nltk.corpus import wordnet, stopwords
from nltk import pos_tag, word_tokenize
import spacy
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('stopwords')

nlp = spacy.load("en_core_web_sm")
```

```python
def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

words = ["studies","am","went","feet","flying","cars","mice","better","ate"]
lemmatizer = WordNetLemmatizer()

for word, tag in pos_tag(words):
    lemma = lemmatizer.lemmatize(word, get_wordnet_pos(tag))
    print(f"{word} ({tag}) → {lemma}")
```

```python
def lemmatize_sentence_list(sent_list):
    results = []
    for sent in sent_list:
        doc = nlp(sent)
        lemma_sent = " ".join([token.lemma_ for token in doc])
        results.append((sent, lemma_sent))
    return results

sentences = ["The students are studying hard.", "He went to the park."]
for orig, lemma in lemmatize_sentence_list(sentences):
    print(f"Original: {orig}\nLemmatized: {lemma}\n")
```

```python
stemmer = PorterStemmer()
words = ["studies","better","flying","easily","running"]

print(f"{'Word':<10}{'Stemmed':<15}{'Lemmatized':<15}")
for w in words:
    stem = stemmer.stem(w)
    lemma = lemmatizer.lemmatize(w, get_wordnet_pos(pos_tag([w])[0][1]))
    print(f"{w:<10}{stem:<15}{lemma:<15}")
```

```python
paragraph = """My university is located in a beautiful city. 
Students from many countries come here to study and share knowledge."""

lemmas = [token.lemma_ for token in nlp(paragraph)]
print(f"Total Words: {len(lemmas)}")
print(f"Unique Lemmas: {len(set(lemmas))}")
print(f"Lemmas: {lemmas}")
```

```python
text = "The dogs were running quickly and happily."
for word, tag in pos_tag(word_tokenize(text)):
    lemma = lemmatizer.lemmatize(word, get_wordnet_pos(tag))
    print(f"{word} ({tag}) → {lemma}")
```

```python
text = "The students are studying hard for their final examinations."
stop_words = set(stopwords.words('english'))
tokens = [w for w in word_tokenize(text.lower()) if w.isalpha() and w not in stop_words]
lemmas = [lemmatizer.lemmatize(w) for w in tokens]
print("Clean tokens:", lemmas)
print("Remaining count:", len(lemmas))
```

```python
text = "Dogs are chasing the cats who were running through the gardens."
nltk_lemmas = [lemmatizer.lemmatize(w, get_wordnet_pos(t)) for w, t in pos_tag(word_tokenize(text))]
spacy_lemmas = [token.lemma_ for token in nlp(text)]
print("NLTK Lemmas:", nltk_lemmas)
print("spaCy Lemmas:", spacy_lemmas)
```

```python
def clean_text(text):
    doc = nlp(text.lower())
    tokens = [t.lemma_ for t in doc if not t.is_stop and t.is_alpha]
    return tokens

text = "Running runners run faster when the weather is better!"
print(clean_text(text))
```

```python
para = """The city is growing rapidly. People are building new houses and improving roads every year."""
lemmas = [token.lemma_.lower() for token in nlp(para) if token.is_alpha]
freq = Counter(lemmas)
print("Top 5 Lemmas:", freq.most_common(5))
```

```python
present = "Students study in the library and read their books."
past = "Students studied in the library and read their books."
present_lemmas = [t.lemma_ for t in nlp(present)]
past_lemmas = [t.lemma_ for t in nlp(past)]
print("Present Lemmas:", present_lemmas)
print("Past Lemmas:", past_lemmas)
```

```python
df = pd.DataFrame({"Review": [
    "The product was amazing and worked perfectly!",
    "I didn’t like the packaging and delivery time."
]})
def lemmatize_reviews(df):
    df["Clean_Review"] = df["Review"].apply(lambda x: " ".join([t.lemma_ for t in nlp(x)]))
    return df
df = lemmatize_reviews(df)
print(df)
```

```python
class MyLemmatizer:
    def __init__(self):
        self.nlp = spacy.load("en_core_web_sm")

    def lemmatize(self, text):
        return [t.lemma_ for t in self.nlp(text) if t.is_alpha]

    def compare(self, s1, s2):
        lem1, lem2 = set(self.lemmatize(s1)), set(self.lemmatize(s2))
        overlap = len(lem1 & lem2) / len(lem1 | lem2) * 100
        print(f"Overlap: {overlap:.2f}%")

ml = MyLemmatizer()
ml.compare("Students study hard in university.", "Pupils studied diligently in college.")
```
