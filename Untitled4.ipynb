```python
import pandas as pd

# Example data
data = {
    "text": [
        "I looove this movie!! 😍 It's sooo good!!",
        "Worst movie ever... I hate it 😡",
        "Just watched it again, still amazing!",
        "Ughhh too slow and boring zzz...",
        "Great acting, nice story, loved it ❤️"
    ],
    "label": [1, 0, 1, 0, 1]  # 1=positive, 0=negative
}

df = pd.DataFrame(data)
df.head()
```

```python
import re
import string

def basic_clean(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)        # remove URLs
    text = re.sub(r"@\w+", "", text)                  # remove mentions
    text = re.sub(r"\d+", "", text)                   # remove digits
    text = text.translate(str.maketrans("", "", string.punctuation))  # remove punctuation
    text = re.sub(r"\s+", " ", text).strip()          # remove extra spaces
    return text

df["text_clean"] = df["text"].apply(basic_clean)
df[["text", "text_clean"]].head()
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter

nltk.download('punkt')
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

# Tokenize
df["tokens"] = df["text_clean"].apply(word_tokenize)

# Remove stopwords
df["tokens_nostop"] = df["tokens"].apply(lambda x: [w for w in x if w not in stop_words])

# Compare top 10 tokens before vs after stopword removal
all_tokens = [w for tokens in df["tokens"] for w in tokens]
all_tokens_nostop = [w for tokens in df["tokens_nostop"] for w in tokens]

print("Top 10 before stopword removal:")
print(Counter(all_tokens).most_common(10))
print("\nTop 10 after stopword removal:")
print(Counter(all_tokens_nostop).most_common(10))
```

```python
from nltk.stem import WordNetLemmatizer, PorterStemmer
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()

sample_words = ["running", "better", "studies", "cars"]

for word in sample_words:
    print(f"{word:10} | Lemma: {lemmatizer.lemmatize(word, pos='v'):10} | Stem: {stemmer.stem(word)}")
```

```python
def normalize_text(text):
    # Reduce repeated characters (e.g., cooool → cool)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    # Simple emoji handling (replace emojis with words)
    text = text.replace("😍", "love").replace("😡", "angry").replace("❤️", "love").replace("zzz", "sleepy")
    return text

df["text_norm"] = df["text_clean"].apply(normalize_text)
df[["text_clean", "text_norm"]].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Bag of Words
bow = CountVectorizer()
X_bow = bow.fit_transform(df["text_norm"])

# TF-IDF
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(df["text_norm"])

print("BoW shape:", X_bow.shape)
print("TF-IDF shape:", X_tfidf.shape)
print("BoW top features:", list(bow.get_feature_names_out())[:10])
print("TF-IDF top features:", list(tfidf.get_feature_names_out())[:10])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Split data
X_train, X_test, y_train, y_test = train_test_split(df["text"], df["label"], test_size=0.3, random_state=42)

# Raw text model
tfidf_raw = TfidfVectorizer()
X_train_raw = tfidf_raw.fit_transform(X_train)
X_test_raw = tfidf_raw.transform(X_test)
model_raw = LogisticRegression()
model_raw.fit(X_train_raw, y_train)
acc_raw = accuracy_score(y_test, model_raw.predict(X_test_raw))

# Cleaned text model
tfidf_clean = TfidfVectorizer()
X_train_clean = tfidf_clean.fit_transform(df.loc[X_train.index, "text_norm"])
X_test_clean = tfidf_clean.transform(df.loc[X_test.index, "text_norm"])
model_clean = LogisticRegression()
model_clean.fit(X_train_clean, y_train)
acc_clean = accuracy_score(y_test, model_clean.predict(X_test_clean))

print(f"Accuracy on raw text: {acc_raw:.2f}")
print(f"Accuracy on cleaned text: {acc_clean:.2f}")
```

```python
df.to_csv("cleaned_text_dataset.csv", index=False)
print("✅ Dataset saved as cleaned_text_dataset.csv")
```
