```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
import string
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
```

```python
raw_text = "I AM Loving the NLP Class 2025, but sometimes it feels Confusing!!!"
```

```python
text_lower = raw_text.lower()
print("Lowercased:", text_lower)
```

```python
#  Tokenization (split into words)
tokens = word_tokenize(text_lower)
print("Tokens:", tokens)
```

```python
#  Remove punctuation & numbers
tokens = [word for word in tokens if word.isalpha()]
print("After removing punctuation/numbers:", tokens)
```

```python
#  Stopword Removal
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]
print("After stopword removal:", filtered_tokens)
```

```python
# Stemming & Lemmatization
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stemmed = [stemmer.stem(word) for word in filtered_tokens]
lemmatized = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("Stemmed:", stemmed)
print("Lemmatized:", lemmatized)
```

```python
#  POS Tagging (on filtered tokens)
pos_tags = pos_tag(filtered_tokens)
print("POS Tags:", pos_tags)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# ---------- Corpus ----------
corpus = [
    "I am loving the NLP class, but sometimes it feels confusing!!!",
    "NLP is a fascinating field — it deals with text, speech, and language understanding."
]
```

```python
# Create CountVectorizer for BOW
vectorizer = CountVectorizer(stop_words='english')  
X = vectorizer.fit_transform(corpus)
```

```python
# Vocabulary (sorted unique words)
vocab = vectorizer.get_feature_names_out()
print("Vocabulary:", vocab)
```

```python
# Bag of Words matrix
bow_matrix = X.toarray()
print("\nBag of Words Matrix:\n", bow_matrix)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
# Corpus
corpus = [
    "I am loving the NLP class, but sometimes it feels confusing!!!",
    "NLP is a fascinating field — it deals with text, speech, and language understanding."
]
```

```python
# Create TF-IDF Vectorizer
vectorizer = TfidfVectorizer(stop_words='english')
```

```python
# Fit and transform the corpus
tfidf_matrix = vectorizer.fit_transform(corpus)
```

```python
# Get vocabulary (features)
vocab = vectorizer.get_feature_names_out()

# Convert to DataFrame for readability
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=vocab)
print(df_tfidf)
```

```python
text = "Call me at +1 (555) 123-4567 or 555-0000."
```

```python
print("Fullmatch:", re.fullmatch(r".*", text).group())  # matches all
```

```python
numbers = re.findall(r"\d+", text)
print("Findall:", numbers)  

for m in re.finditer(r"\d+", text):
    print("Found:", m.group(), "at", m.span())
```

```python
print(re.findall(r"[aeiou]", "Hello World"))   
print(re.findall(r"[^aeiou]", "Hello World"))  
print(re.findall(r"\d", "a1b2c3"))            
print(re.findall(r"\w", "a1_b2"))              
print(re.findall(r"\s", "a b\tc\n"))           

```

```python
s = "<p>first</p><p>second</p>"

print("Greedy:", re.findall(r"<p>.*</p>", s))   # one big match
print("Lazy:", re.findall(r"<p>.*?</p>", s))    # separate matches
```

```python
date = "5/11/2024"
m = re.search(r"(\d{1,2})/(\d{1,2})/(\d{4})", date)
print("Day:", m.group(1))   
print("Month:", m.group(2)) 
print("Year:", m.group(3))  
```

```python
m = re.search(r"(?P<day>\d{1,2})/(?P<month>\d{1,2})/(?P<year>\d{4})", date)
print("Named groups:", m.groupdict())
```

```python
text = "20kg 30 40kg 50"
print(re.findall(r"\d+(?!kg)", text))   
print(re.findall(r"\d+(?=kg)", text))  
```

```python
text = "Python\njava\nC++"
print(re.findall(r"^.+", text))               # only first line
print(re.findall(r"^.+", text, re.M))         # all lines (multiline)
print(re.findall(r"python", text, re.I))      # case-insensitive
pattern = re.compile(r"""
    \d+     # number
    \s*     # optional spaces
    kg      # literal text
""", re.X)   
print(pattern.findall("20kg 30 kg"))
```

```python
def double(m):
    return str(int(m.group()) * 2)
print(re.sub(r"\d+", double, "a1b2c3"))  

```
